## 2. Conferir a base tratada

```python
import pandas as pd

output_path = PROJECT_ROOT / "data" / "processed" / "censo_escolar_2024_municipal_ativa_anos_iniciais_sem_nulos.csv"
data = pd.read_csv(output_path, sep=";", encoding="utf-8-sig")

checks = {
    "linhas": len(data),
    "nulos_banda_larga": int(data["IN_BANDA_LARGA"].isna().sum()),
    "valores_banda_larga": sorted(data["IN_BANDA_LARGA"].unique().tolist()),
    "sem_internet_com_banda": int(((data["IN_INTERNET"] == 0) & (data["IN_BANDA_LARGA"] == 1)).sum()),
}
checks
```
